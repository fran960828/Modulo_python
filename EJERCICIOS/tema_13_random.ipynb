### 8. Simular un sorteo

Mezcla 200 boletos y elige 3 ganadores y 2 suplentes, todos distintos.

```python
boletos=[num for num in range(1,201)]
rd.shuffle(boletos)
print(boletos)
ganadores=rd.sample(boletos,k=3)
print(ganadores)
boletos_restantes=[num for num in boletos if num not in ganadores]
suplentes=rd.sample(boletos_restantes,k=2)
print(suplentes)
```


### 10. Permutación aleatoria controlada

Asigna asientos al azar a 30 personas tres veces y verifica si alguien
repite asiento.

```python

lista_personas = [f'persona {num}' for num in range(1,31)]
lista_asientos = [f'Asiento {i}' for i in range(1,31)]

asignaciones = []

# Generamos 3 asignaciones
for _ in range(3):
    asientos_aleatorios = lista_asientos.copy()
    rd.shuffle(asientos_aleatorios)
    asignaciones.append(dict(zip(lista_personas, asientos_aleatorios)))

# Verificar repeticiones de asiento por persona
for persona in (lista_personas):
    asientos_persona = [asig[persona] for asig in asignaciones]
    if len(set(asientos_persona)) < 3:
        print(f"{persona} repitió asiento en alguna permutación: {asientos_persona}")
```

### 11. Transacciones bancarias falsas

Genera 100 transacciones ficticias con ID único, cliente, monto, tipo,
ciudad y fecha.

```python
from datetime import date
import uuid
```

```python


# Generar datos
clientes = [f'cliente {num}' for num in range(1, 101)]
montos = [rd.randint(500, 5000) for _ in range(100)]
tipos = rd.choices(['tarjeta','efectivo','transferencia'], k=100)
ciudades = rd.choices(['Madrid','Barcelona','Valencia'], k=100)

fechas = []
for _ in range(100):
    day = rd.randint(1, 28)
    mes = rd.randint(1, 12)
    año = 2025
    fechas.append(date(año, mes, day))

# Generar transacciones con ID único
transacciones = {}
for i in range(100):
    id_unico = uuid.uuid4()
    transacciones[id_unico] = {
        "cliente": clientes[i],
        "monto": montos[i],
        "tipo": tipos[i],
        "ciudad": ciudades[i],
        "fecha": fechas[i].isoformat()
    }

# Mostrar ejemplo de 5 transacciones
for i, (id_tx, tx) in enumerate(transacciones.items()):
    print(f"{id_tx}: {tx}")
    if i >= 4:
        break

```

### 13. Generador de perfiles

Crea 500 perfiles con campos aleatorios como edad, e‑mail, ciudad,
preferencias e historial de pagos.

```python
import string
```

```python

# ---------- Funciones auxiliares profesionales ----------

def generar_email():
    """Genera un email único ficticio."""
    abecedario = string.ascii_lowercase + string.digits
    nombre = ''.join(rd.choices(abecedario, k=10))
    dominios = ["gmail.com", "hotmail.com", "yahoo.com"]
    return f"{nombre}@{rd.choice(dominios)}"
```

```python

# ---------- Funciones auxiliares profesionales ----------

def generar_email():
    """Genera un email único ficticio."""
    abecedario = string.ascii_lowercase + string.digits
    nombre = ''.join(rd.choices(abecedario, k=10))
    dominios = ["gmail.com", "hotmail.com", "yahoo.com"]
    return f"{nombre}@{rd.choice(dominios)}"


def generar_historial_pagos():
    """Crea un historial de pagos realista para una persona."""
    historial = []
    n_pagos = rd.randint(1, 12)   # 1 a 12 pagos (mensual)
    
    for _ in range(n_pagos):
        monto = rd.randint(15, 500)

        # Fecha aleatoria entre 2024 y 2025
        año = rd.choice([2024, 2025])
        mes = rd.randint(1, 12)
        dia = rd.randint(1, 28)
        fecha = date(año, mes, dia)

        historial.append({"fecha": fecha, "monto": monto})

    return historial


def generar_preferencias():
    """Devuelve 2 preferencias distintas."""
    opciones = ['Deporte', 'Viajar', 'Trabajar', 'Netflix', 'Cocina', 'Lectura']
    return rd.sample(opciones, k=2)


def generar_nombre():
    """Genera un nombre ficticio simple."""
    nombres = ["Ana", "Luis", "Carlos", "María", "Sofía", "Pedro", "Laura", "David", "Lucía", "Pablo"]
    apellidos = ["García", "López", "Martínez", "Sánchez", "Romero", "Muñoz", "Navarro"]
    return f"{rd.choice(nombres)} {rd.choice(apellidos)}"


# ---------- Generación de los perfiles ----------

perfiles = {}

for i in range(500):
    perfiles[f"perfil_{i+1}"] = {
        "nombre": generar_nombre(),
        "edad": rd.randint(18, 80),
        "email": generar_email(),
        "ciudad": rd.choice(["Madrid", "Valencia", "Zaragoza", "Barcelona"]),
        "preferencias": generar_preferencias(),
        "historial_pagos": generar_historial_pagos()
    }

perfiles

```
