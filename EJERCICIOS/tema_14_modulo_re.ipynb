### **Ejercicio 1 -- Coincidencia literal sencilla**

Encuentra todas las ocurrencias exactas del texto:

    INFO

dentro del siguiente contenido:

    INFO: Inicio del proceso
    Error: información no disponible
    UserINFO: registro completo
    INFO

```python
import re
```

```python
def find_matches(pattern_compiled, text):
    """Ejecuta la búsqueda e imprime los objetos match encontrados."""
    matches = pattern_compiled.finditer(text)
    # Mostramos el patrón compilado para referencia
    print(f"\n--- Patrón: '{pattern_compiled.pattern}' ---")
    for match in matches:
        print(match)
```

```python
text_ej1='''INFO: Inicio del proceso
    Error: información no disponible
    UserINFO: registro completo
    INFO'''
```

```python
# Definimos el patron que estamos buscando
pattern_ej1=re.compile('INFO')
# Ejecutamos la función de busqueda que nos da las posiciones donde se encuentran las coincidencias
find_matches(pattern_ej1,text_ej1)
```

Detecta todas las cadenas que contengan el símbolo literal `+`:

    3+5=8
    Hello+World
    Increment++
    C++ language
    A+B no se captura

Capta únicamente las líneas donde el símbolo `+` aparece **literalmente
al menos una vez**, ignorando las que contienen `++`.

```python
text_ej2='''3+5=8
    Hello+World
    Increment++
    C++ language
    A+B no se captura'''
```

```python
pattern_ej2=re.compile(r'\+\w')
resultado_ej2=[frase.strip() for frase in text_ej2.split('\n') if pattern_ej2.search(frase) is not None ]
for frase in resultado_ej2:
    print(frase)
```

### **Ejercicio 13 -- Captura de correos**

Extrae usuario, dominio y TLD:

    juan23@mail.com
    maria.santos@empresa.org
    root@localhost.local
    ventas-online@tienda.es

```python

pattern = re.compile(r'\b(\w+[-.]?\w+?)(@\w+)([.]\w+)')

emails = """
juan23@mail.com
maria.santos@empresa.org
root@localhost.local
ventas-online@tienda.es
"""

for match in pattern.finditer(emails):
    print(match.group(1))
    print(match.group(2))
    print(match.group(3))
    
```

### **Ejercicio 15 -- Sustitución de fechas**

Convierte dd/mm/yyyy → yyyy-mm-dd:

    23/01/2024
    5/12/2023
    01/01/2025

```python
text_ej15='''23/01/2024
    5/12/2023
    01/01/2025'''
pattern_ej15=re.compile(r'([0-9]?[0-9])[/]([0-9][0-9])[/]([0-9][0-9][0-9][0-9])')
for match in pattern_ej15.finditer(text_ej15):
    print(f'{match.group(3)}-{match.group(2)}-{match.group(1)}')
```

### **Ejercicio 20 -- Análisis de logs**

Extrae: - fecha - hora - nivel - módulo - mensaje

    [2025-01-05 11:33:21] ERROR (moduleA): File not found
    [2025-01-05 11:33:25] INFO (moduleB): Started process
    [2025-01-05 11:33:40] WARNING (core): Low memory

```python
text_ej20='''
    [2025-01-05 11:33:21] ERROR (moduleA): File not found
    [2025-01-05 11:33:25] INFO (moduleB): Started process
    [2025-01-05 11:33:40] WARNING (core): Low memory'''
```

```python
pattern_ej20=re.compile(r'\[(\d{4}\-\d{2}\-\d{2})\s(\d{2}:\d{2}:\d{2})\]\s+(\w+)\s+\(([^)]+)\):\s+(.*)')
```

```python
for m in pattern_ej20.finditer(text_ej20):
    fecha = m.group(1)
    hora = m.group(2)
    problema = m.group(3)
    modulo = m.group(4)
    mensaje = m.group(5)

    print("Fecha:", fecha)
    print("Hora:", hora)
    print("Problema:", problema)
    print("Módulo:", modulo)
    print("Mensaje:", mensaje)
    print("-" * 40)
```

### **Ejercicio 22 -- Normalizar URLs**

Transforma:

    HTTPS://Example.COM/page
    http://blog.example.com
    www.Example.org

En versiones normalizadas con: - https:// obligatorio - minúsculas

```python
text_ej22='''HTTPS://Example.COM/page
    http://blog.example.com
    www.Example.org'''
```

```python
pattern_ej22 = re.compile(r'(?i)(?:https?://)?(?=www\.)?([\w.-]+\.\w+)')
```

```python
for m in pattern_ej22.finditer(text_ej22):
    print(f'https://{m.group(1).lower()}')
```
