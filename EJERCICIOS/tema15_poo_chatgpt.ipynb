✅ Ejercicio 7: Herencia básica

Crea una clase base Animal:

name

Método make_sound()

Crea subclases:

Dog

Cat

Cada uno debe sobrescribir make_sound().

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    def __init__(self, name):
        self.name = name

    @abstractmethod
    def make_sound(self):
        pass


class Dog(Animal):
    def make_sound(self):
        return f'El {self.name} hace guau'


class Cat(Animal):
    def make_sound(self):
        return f'El {self.name} hace miau'


animal_1 = Dog('perro')
animal_2 = Cat('gato')

print(animal_1.make_sound())
print(animal_2.make_sound())
```

✅ Ejercicio 8: Sistema de empleados

Crea:

Employee: nombre, salario

SalesEmployee: hereda de Employee y tiene commission

TechEmployee: hereda de Employee y tiene level

Cada clase debe tener su propio método get_salary().

```python
class Employee(ABC):
    def __init__(self,nombre,salario):
        super().__init__()
        self.nombre=nombre
        self.salario=salario
    
    @abstractmethod
    def get_salary(self):
        pass

class SalesEmployee(Employee):
    def __init__(self, nombre, salario,comision):
        super().__init__(nombre, salario)
        self.comision=comision
    def get_salary(self):
        self.salario*=self.comision
        print(self.salario)
    
class TechEmployee(Employee):
    def __init__(self, nombre, salario,level):
        super().__init__(nombre, salario)
        self.level=level
    def get_salary(self):
        if self.level=='junior':
            print(self.salario)
        elif self.level=='mid-level':
            self.salario*=1.7
            print(self.salario*1.7)
        elif self.level=='senior':
            self.salario*=2.5
            print(self.salario*2.5)
```

```python
from abc import ABC, abstractmethod

class Employee(ABC):
    def __init__(self, nombre, salario):
        self.nombre = nombre
        self.salario = salario

    @abstractmethod
    def get_salary(self):
        pass


class SalesEmployee(Employee):
    def __init__(self, nombre, salario, comision):
        super().__init__(nombre, salario)
        self.comision = comision
    def __str__(self):
        return f'{self.nombre} tiene un salario de {self.salario}'
    def get_salary(self):
        return self.salario * self.comision   # ✅ devuelve, no modifica


class TechEmployee(Employee):
    def __init__(self, nombre, salario, level,prog_language):
        super().__init__(nombre, salario)
        self.level = level
        self.prog_language=prog_language
    def __str__(self):
        return f'{self.nombre} tiene un salario de {self.salario} y domina el lenguage de {self.prog_language}'

    def get_salary(self):
        if self.level == 1:
            return self.salario
        elif self.level == 2:
            return self.salario * 1.7
        elif self.level == 3:
            return self.salario * 2.5
        else:
            raise ValueError("Nivel técnico inválido")
    
```

```python
salesman1=SalesEmployee('Mari',1800,1.05)
print(salesman1.get_salary())
developer1=TechEmployee('Fran',3000,2,'java')
print(developer1.get_salary())
print(developer1)
```

✅ Ejercicio 10: Clase TeamLeader

Crea:

Clase TeamLeader que herede de Employee

Debe manejar una lista de empleados a su cargo

Debe tener métodos:

add_employee()

remove_employee()

show_team()

```python
class TeamLeader(Employee):
    def __init__(self, nombre, salario):
        super().__init__(nombre, salario)
        self.empleados = []
    def get_salary(self):
        return super().get_salary()

    def add_employee(self, empl):
        if empl not in self.empleados:
            self.empleados.append(empl)
        else:
            raise ValueError('El empleado ya está en la base de datos')

    def remove_employee(self, empl):
        if empl in self.empleados:
            self.empleados.remove(empl)
        else:
            raise ValueError('El empleado no se encuentra en la base de datos')

    def show_team(self):
        if len(self.empleados) > 0:
            print(f'Empleados a cargo de {self.nombre}:')
            for empl in self.empleados:
                print(empl)
        else:
            raise ValueError(f'{self.nombre} no tiene empleados a cargo')
```

```python
manager=TeamLeader('Arturo',5000)
manager.add_employee(salesman1)
manager.add_employee(developer1)
manager.show_team()
manager.remove_employee(salesman1)
manager.show_team()
```

```python
manager2=TeamLeader('Ivan',6000)
manager2.add_employee(salesman1)
manager2.show_team()
```

✅ Ejercicio 15: Getter automático

Crea una clase User:

name, last

Propiedad email calculada automáticamente

✅ Ejercicio 16: Setter

Usa @property para:

Permitir asignar full_name = "Ana Lopez"

Que actualice nombre y apellido

✅ Ejercicio 17: Deleter

Cuando se haga:

del user.full_name


Debe:

Borrar los datos

Imprimir "Usuario eliminado"

```python

class User:
    def __init__(self,name,last):
        self.name=name
        self.last=last
    @property
    def email(self):
        return f'{self.name}.{self.last}@hotmail.com'
    @property
    def fullname(self):
        return f'{self.name} {self.last}'
    @fullname.setter
    def fullname(self, fullname):
        partes = fullname.split(' ')
        if len(partes) != 2:
            raise ValueError("El nombre completo debe contener exactamente nombre y apellido")
        self.name, self.last = partes
    @fullname.deleter
    def fullname(self):
        self.name=None
        self.last=None
```

```python
user1=User('Fran','ng96')
#Uso de getter
print(user1.email)
user1.name='Francisco'
print(user1.email)
#Fullname al usar property para get, setter o deleter pasa a ser un atributo y se llama como tal, da error si lo llamas como una función
user1.fullname='Francisco Navarro'
print(user1.email)
# uso de deleter
del user1.fullname
print(user1.email)
```

✅ Ejercicio 19: Constructor alternativo

Crea una clase Client que acepte:

"Pedro;Gomez;35"


como entrada usando un @classmethod.

```python
class Client:
    def __init__(self,nombre,apellido,edad):
        self.nombre=nombre 
        self.apellido=apellido 
        self.edad=edad 
    def __str__(self):
        return f'{self.nombre} {self.apellido} tiene {self.edad} años'
    @classmethod
    def constructor_clase(cls,client:str):
        nombre,apellido,edad=client.split(';')
        edad=int(edad)
        return cls(nombre,apellido,edad)

cliente1=Client.constructor_clase('Pedro;Gomez;44')
print(cliente1)
```


✅ Ejercicio 20: Static Method utilitario

Crea una clase Temperature con:

Static method celsius_to_fahrenheit()

Static method fahrenheit_to_celsius()

```python
class Temperature:

    @staticmethod
    def celsius_to_fahrenheit(grados_celsius: float):
        return (grados_celsius * 9/5) + 32

    @staticmethod
    def farenheit_to_celsius(grados_farenheit: float):
        return (grados_farenheit - 32) * 5/9


print(Temperature.celsius_to_fahrenheit(24))   # 75.2
print(Temperature.farenheit_to_celsius(104))  # 40.0
```
