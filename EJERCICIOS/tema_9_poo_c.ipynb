Imagina que estás desarrollando un sistema de reservas de vuelos para una
aerolínea. Crea un sistema de clases que permita a los usuarios realizar
reservas de vuelos. Aquí tienes una posible estructura:
- Clase base: `Vuelo`
- Atributos: número de vuelo, origen, destino, capacidad máxima, lista de
pasajeros
- Métodos: agregar pasajero, verificar disponibilidad de asientos
- Clase derivada: `VueloEspecial` (hereda de `Vuelo`)
- Atributos adicionales: motivo del vuelo especial (por ejemplo, vacaciones,
trabajo)
Resuelve el problema creando instancias de estas clases y realizando
reservas para diferentes vuelos y tipos de vuelos especiales.

```python
class Vuelo:
    def __init__(self,numeroVuelo,origen,destino,capacidadMax,listaPasajeros):
        self.numeroVuelo=numeroVuelo
        self.origen=origen
        self.destino=destino
        self.capacidadMax=capacidadMax
        self.listaPasajeros=listaPasajeros
    
    def __str__(self):
        return f'Numero de vuelo {self.numeroVuelo} con destino {self.destino} y numero de pasajeros {self.listaPasajeros}'

    def verificar_disponibilidad(self):
        return self.capacidadMax>self.listaPasajeros
    def agregar_pasajeros(self,boletos):
        if self.verificar_disponibilidad() and boletos<=(self.capacidadMax-self.listaPasajeros):
            self.listaPasajeros+=boletos
            
```

```python
vuelo1=Vuelo('123456789','Madrid','Paris',250,0)
vuelo1.agregar_pasajeros(25)
vuelo1.agregar_pasajeros(225)
print(vuelo1.verificar_disponibilidad())
print(vuelo1)
```

```python
class VueloEspecial(Vuelo):
    def __init__(self,numeroVuelo,origen,destino,capacidadMax,listaPasajeros,motivo):
        super().__init__(numeroVuelo,origen,destino,capacidadMax,listaPasajeros)
```

```python
vueloespecial1=VueloEspecial('12345432','Madrid','Berlin',250,0,'Vacaciones')
vueloespecial1.agregar_pasajeros(100)
print(vueloespecial1)
```

Supongamos que estás construyendo un sistema de gestión de empleados
para una empresa. Crea un sistema de clases que maneje la información de
los empleados, incluyendo empleados a tiempo completo y empleados a
tiempo parcial.
- Clase base: `Empleado`
- Atributos: nombre, apellido, salario base
- Clase derivada: `EmpleadoTiempoCompleto` (hereda de `Empleado`)
- Atributo adicional: bono anual
- Clase derivada: `EmpleadoTiempoParcial` (hereda de `Empleado`)
- Atributo adicional: horas trabajadas por semana
Resuelve el problema creando instancias de estas clases y calculando los
salarios totales para diferentes tipos de empleados.

```python
class Empleado:
    def __init__(self,nombre,apellido):
        self.nombre=nombre
        self.apellido=apellido
        
```

```python
class EmpleadoTiempoCompleto(Empleado):
    def __init__(self,nombre,apellido,salario_base,bono_anual):
        super().__init__(nombre,apellido)
        self.salario_base=salario_base
        self.bono_anual=bono_anual
    
    def salario_total(self):
        return self.salario_base*12+self.bono_anual
    
```

```python
class EmpleadoTiempoParcial(Empleado):
    def __init__(self,nombre,apellido,horas_totales):
        super().__init__(nombre,apellido)
        self.horas_totales=horas_totales
    def salario_parcial(self):
        return self.horas_totales*8*52
```

```python
empleadofijo=EmpleadoTiempoCompleto('Francisco','Navarro',1250,2700)
empleado_parcial=EmpleadoTiempoParcial('Jose','Navarro',40)
print(empleadofijo.salario_total())
print(empleado_parcial.salario_parcial())
```
