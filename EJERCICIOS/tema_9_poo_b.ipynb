Crea una clase "CuentaBancaria" con atributos como número de cuenta y 
saldo. Implementa métodos para depositar y retirar dinero, y muestra el 
saldo actual.

```python
class CuentaBancaria:
    def __init__(self,num_cuenta,saldo):
        self.num_cuenta=num_cuenta
        self.saldo=saldo
    def __str__(self):
        return f'el saldo actual es {self.saldo}'
    def depositar_dinero(self,deposito):
        try:
            self.saldo+=deposito
        except Exception as e:
            print(f'Ingreso no valido: {e}')
    def retirar_dinero(self,retiro):
        try:
            if retiro<=self.saldo:
                self.saldo-=retiro
            else:
                print('Tu saldo actual no cubre el retiro')
        except Exception as e:
            print(f'Retiro no valido {e}')    
    
```

Crea una clase "Producto" con atributos como nombre, precio y cantidad en 
stock. Luego, crea una clase "Tienda" que contenga una lista de productos 
disponibles y métodos para agregar productos, mostrar el inventario y 
realizar una compra.

```python
class Producto:
    """Representa un producto con nombre, precio y cantidad disponible."""
    def __init__(self, nombre: str, precio: float, cantidad: int):
        self.nombre = nombre
        self.precio = precio
        self.cantidad = cantidad

    def __str__(self):
        return f"{self.nombre} (Precio: {self.precio}, Cantidad: {self.cantidad})"

    def agregar_cantidad(self, cantidad: int):
        """Aumenta el stock del producto."""
        self.cantidad += cantidad

    def reducir_cantidad(self, cantidad: int):
        """Disminuye el stock del producto, validando disponibilidad."""
        if cantidad > self.cantidad:
            raise ValueError("Cantidad insuficiente en inventario.")
        self.cantidad -= cantidad

    def sin_stock(self) -> bool:
        """Indica si el producto se agotó."""
        return self.cantidad == 0


class Tienda:
    """Gestiona el inventario de productos."""
    def __init__(self):
        self._stock = {}

    def agregar_producto(self, producto: Producto):
        """Agrega un nuevo producto o actualiza su cantidad."""
        if producto.nombre in self._stock:#lista de keys por defecto por eso usa producto.nombre en lugar de solo producto
            self._stock[producto.nombre].agregar_cantidad(producto.cantidad)
            print(f"Cantidad de '{producto.nombre}' actualizada.")
        else:
            self._stock[producto.nombre] = producto
            print(f"'{producto.nombre}' ha sido agregado correctamente.")

    def mostrar_inventario(self):
        """Muestra todos los productos en el inventario."""
        if not self._stock:
            print("El inventario está vacío.")
            return
        for producto in self._stock.values():
            print(f"Producto: {producto.nombre} -> Precio: {producto.precio} -> Cantidad: {producto.cantidad}")

    def comprar_producto(self, nombre_producto: str, cantidad: int):
        """Permite la compra de un producto si hay suficiente cantidad."""
        producto = self._stock.get(nombre_producto)

        if not producto:
            print(f"Producto '{nombre_producto}' no encontrado.")
            return

        try:
            producto.reducir_cantidad(cantidad)
            print(f"Compra de '{nombre_producto}' con cantidad {cantidad} realizada.")
            if producto.sin_stock():
                del self._stock[nombre_producto]
                print(f"'{nombre_producto}' se ha agotado y fue removido del inventario.")
        except ValueError:
            print(f"Cantidad de '{nombre_producto}' insuficiente.")
```

```python
# Crear productos
p1 = Producto("Manzana", 1.0, 10)
p2 = Producto("Banana", 0.5, 20)

# Crear tienda
tienda = Tienda()
tienda.agregar_producto(p1)
tienda.agregar_producto(p2)

# Mostrar inventario
tienda.mostrar_inventario()

# Comprar producto
tienda.comprar_producto("Manzana", 5)
tienda.comprar_producto("Manzana", 5)  # Se elimina al quedar sin stock
tienda.mostrar_inventario()
```

En programación, un "stack" es una estructura de datos que sigue el 
principio de LIFO (Last In, First Out), lo que significa que el último elemento 
agregado a la pila es el primero en ser retirado. Imagina una pila de platos: 
cuando apilas un nuevo plato, este se coloca en la parte superior de la pila, 
y cuando retiras un plato, siempre tomas el de arriba primero.
 En Python, puedes implementar un stack utilizando una lista. Puedes 
agregar elementos a la pila utilizando el método `append()`, y puedes retirar 
elementos de la pila utilizando el método `pop()` sin ningún índice 
especificado, ya que `pop()` por defecto elimina y devuelve el último 
elemento de la lista.
Los stacks son útiles en muchas situaciones, como algoritmos de búsqueda 
y recorrido, manejo de llamadas a funciones (con la pila de llamadas), 
manejos de historial y navegación y más.
 Crea una clase "Pila" que represente una pila (stack) básica. Implementa 
métodos para agregar elementos a la pila, quitar elementos y mostrar el 
contenido actual.
 Por supuesto, estaré encantado de explicarte qué es un "stack" en el 
contexto de la programación y cómo se utiliza en Python.

```python
class Pila:
    """Clase que representa una pila (stack) siguiendo el principio LIFO."""
    
    def __init__(self):
        self._elementos = []  # El guion bajo indica que es un atributo "interno"

    def apilar(self, elemento):
        """Agrega un elemento encima de la pila."""
        self._elementos.append(elemento)
        print(f"Elemento '{elemento}' agregado a la pila.")

    def desapilar(self):
        """Elimina y devuelve el último elemento agregado (el de arriba)."""
        if self.esta_vacia():
            print("La pila está vacía, no se puede desapilar.")
            return None
        elemento = self._elementos.pop()
        print(f"Elemento '{elemento}' removido de la pila.")
        return elemento

    def mostrar(self):
        """Muestra el contenido actual de la pila."""
        if self.esta_vacia():
            print("La pila está vacía.")
        else:
            print("Contenido de la pila (de abajo hacia arriba):")
            for elemento in self._elementos:
                print(f" - {elemento}")

    def esta_vacia(self):
        """Verifica si la pila está vacía."""
        return len(self._elementos) == 0

    def cima(self):
        """Devuelve el elemento que está en la cima de la pila sin quitarlo."""
        if self.esta_vacia():
            print("La pila está vacía.")
            return None
        return self._elementos[-1]

```
