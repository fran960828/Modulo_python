 Problema de Optimización de Subarreglo:
 Imagina que estás trabajando en un sistema de análisis de datos y te han 
proporcionado una lista de números enteros. Tu tarea es desarrollar una 
función llamada 
max_subarray_sum que encuentre y devuelva la suma máxima de 
un subarreglo contiguo en la lista.
 Por ejemplo, considera la lista 
[1, -2, 3, 10, -4, 7, 2, -5] . El subarreglo 
contiguo con la suma máxima es 
[3, 10, -4, 7, 2] , y la suma de esos elementos 
es 18. Por lo tanto, la función debería devolver 
18 para esta lista.
 Implementa la función 
max_subarray_sum y, además, aplica memoización para 
mejorar su eficiencia en el cálculo de subarreglos de suma máxima en listas 
previamente procesadas.

```python
from functools import lru_cache
import time
```

```python
lista=[1, -2, 3, 10, -4, 7, 2, -5]
subarray=[]
def max_subarray_sum(array,index=0,max_actual=0,max_global=float('-inf')):
    # Condición base
    if len(array) == index:
        return max_global
    max_actual=max(array[index],max_actual+array[index])
    max_global=max(max_actual,max_global)
    if max_global==max_actual:
        subarray.append(array[index])
    return max_subarray_sum(array,index+1,max_actual,max_global)
```

```python
max_subarray_sum(lista)
```

```python
from functools import lru_cache

def max_cross_sum(array, left, mid, right):
    """Suma máxima que cruza el punto medio"""
    left_sum = float('-inf')
    right_sum = float('-inf')
    temp = 0

    # Parte izquierda (desde mid hacia la izquierda)
    for i in range(mid, left - 1, -1):
        temp += array[i]
        left_sum = max(left_sum, temp)

    temp = 0
    # Parte derecha (desde mid+1 hacia la derecha)
    for i in range(mid + 1, right + 1):
        temp += array[i]
        right_sum = max(right_sum, temp)

    return left_sum + right_sum

@lru_cache(maxsize=None)
def max_subarray_memo(array_tuple, left, right):
    """Versión recursiva con memoización (array inmutable como tuple)"""
    if left == right:
        return array_tuple[left]

    mid = (left + right) // 2
    left_sum = max_subarray_memo(array_tuple, left, mid)
    right_sum = max_subarray_memo(array_tuple, mid + 1, right)
    cross_sum = max_cross_sum(array_tuple, left, mid, right)

    return max(left_sum, right_sum, cross_sum)
```

Imagina que estás trabajando en un sistema de gestión de costos de envío para 
una empresa de logística. El sistema debe calcular el costo de envío para 
diferentes destinos, distancias y pesos de paquetes. Implementa una función 
llamada 
calcular_costo_envio que tome como entrada un destino, una distancia 
en kilómetros y un peso en kilogramos, y devuelva el costo total del envío.
 Requerimientos:
  El costo base de envío es de $5.0.
  El costo por kilómetro de distancia es de $0.1.
  El costo por kilogramo de peso es de $0.2.
 Implementa la función de manera eficiente utilizando memoización para evitar 
recalcular el costo para los mismos destinos, distancias y pesos.

```python
from functools import lru_cache
import time
```

```python
@lru_cache(maxsize=None)
def calcular_costo_envio(lugar,distancia_km,peso_kg):
    time.sleep(1)
    costo_base=5
    costo_total=costo_base+distancia_km*0.1+peso_kg*0.2
    return {lugar:costo_total}
```

```python
inicio=time.time()
coste=calcular_costo_envio('paris',1500,6)
fin=time.time()
print(coste)
print(fin-inicio)
```

Imagina que estás trabajando en un sistema de análisis de texto que requiere 
calcular la frecuencia de ocurrencia de palabras en un conjunto de 
documentos. Implementa una función llamada 
calcular_frecuencia_palabras que 
tome como entrada un texto y devuelva un diccionario que muestre la 
frecuencia de cada palabra en el texto.
  La función debe ser capaz de manejar textos y ser insensible a 
mayúsculas/minúsculas (por ejemplo, "Hola" y "hola" se consideran la 
misma palabra).
  Se deben excluir las palabras comunes (artículos, preposiciones, etc.) que 
no aportan información relevante al análisis.
 Utiliza memoización para evitar recalcular la frecuencia de palabras para el 
mismo texto.

```python
try:
    import nltk
    from nltk.corpus import stopwords
    from nltk.tokenize import word_tokenize
    import re
    import unicodedata
    from collections import Counter
    nltk.download('punkt_tab')
    nltk.download('stopwords')
    from functools import lru_cache
except ModuleNotFoundError:
    print('modulo no encontrado')
```

```python
texto='''Durante los últimos años, muchas empresas han comenzado a reconocer la importancia de la sostenibilidad y la responsabilidad ambiental en sus operaciones diarias. Sin embargo, no todas han logrado implementar estrategias efectivas que realmente reduzcan su impacto en el medio ambiente. Algunas compañías simplemente utilizan términos como “verde” o “ecológico” con fines publicitarios, sin realizar cambios significativos en sus procesos productivos.

Por otro lado, existen organizaciones que sí han adoptado medidas concretas: optimizan el consumo energético, reducen los residuos y promueven el uso de materiales reciclables. Estas acciones, aunque parezcan pequeñas, pueden marcar una gran diferencia a largo plazo. Además, los consumidores están cada vez más informados y tienden a preferir marcas que demuestren un compromiso auténtico con el planeta.

No obstante, el cambio hacia una economía más sostenible no depende únicamente de las empresas. Los gobiernos deben establecer políticas claras, ofrecer incentivos fiscales y fomentar la educación ambiental desde las escuelas. También los ciudadanos tienen un papel esencial: reducir el consumo innecesario, reutilizar productos y separar correctamente los desechos.

En definitiva, construir un futuro más limpio y equilibrado requiere la colaboración de todos los sectores de la sociedad. Solo mediante un esfuerzo conjunto podremos asegurar que las próximas generaciones disfruten de un entorno saludable, justo y próspero.'''
```

```python
# --- Función para limpiar el texto ---
def limpiar_texto(texto):
    """
    Limpia el texto recibido:
    1. Convierte todo el texto a minúsculas.
    2. Elimina tildes y diacríticos usando unicodedata.
    3. Elimina cualquier carácter que no sea letra (incluye letras acentuadas y la ñ) o espacio.
    """
    texto = texto.lower()  # Convierte a minúsculas para uniformidad
    # Normaliza el texto y elimina los caracteres de tipo 'Mn' (marcas de acento)
    texto = ''.join(c for c in unicodedata.normalize('NFD', texto)
                    if unicodedata.category(c) != 'Mn')  # quita tildes
    # Elimina cualquier carácter que no sea letra o espacio
    texto = re.sub(r'[^a-záéíóúüñ\s]', '', texto)
    return texto

# --- Función para eliminar palabras vacías (stopwords) ---
def eliminar_stopword(texto):
    """
    Tokeniza el texto y elimina las palabras vacías (stopwords) en español.
    Solo conserva palabras alfabéticas que no estén en la lista de stopwords.
    """
    tokens = word_tokenize(texto, language='spanish', preserve_line=False)  # Tokeniza el texto
    # Filtra tokens: solo palabras alfabéticas y que no sean stopwords
    tokens_limpios = [palabra for palabra in tokens if palabra.isalpha() and palabra not in stopwords.words('spanish')]
    return tokens_limpios

# --- Función para contar la frecuencia de cada palabra ---
@lru_cache(maxsize=None)
def contador(tokens):
    """
    Cuenta la frecuencia de cada palabra en la lista de tokens usando Counter.
    Devuelve un diccionario con la frecuencia de cada palabra.
    """
    frecuencia = Counter(tokens)
    return frecuencia

# --- Función para guardar el resultado en un archivo ---
def guardar_resultado(conteo, nombre_archivo="resultado.txt"):
    """
    Guarda el conteo de palabras en un archivo de texto.
    Cada línea contiene una palabra y su frecuencia.
    """
    with open(nombre_archivo, "w", encoding="utf-8") as archivo:
        for palabra, frecuencia in conteo.items():
            archivo.write(f"{palabra}: {frecuencia}\n")

# --- Función para guardar errores en un archivo ---
def guardar_error(error, nombre_archivo="error_log.txt"):
    """
    Guarda el mensaje de error en un archivo de texto.
    :param error: Excepción capturada (objeto Exception)
    :param nombre_archivo: nombre del archivo donde se guardará el error
    """
    with open(nombre_archivo, "w", encoding="utf-8") as archivo:
        archivo.write(f"{type(error).__name__}: {error}\n")

# --- Flujo principal de procesamiento ---
try:
    texto_limpio = limpiar_texto(texto)  # Limpia el texto original
    texto_stopwords = eliminar_stopword(texto_limpio)  # Elimina stopwords
    recuento = contador(texto_stopwords)  # Cuenta la frecuencia de palabras
except Exception as e:
    guardar_error(e)  # Si ocurre un error, lo guarda en un archivo
else:
    guardar_resultado(recuento)  # Si todo va bien, guarda el resultado en un archivo
```
