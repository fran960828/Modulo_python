Fase 3: Raspado de Múltiples Elementos (Bucle y find_all)
Para extraer toda la información relevante de la página (no solo el primer resultado), es necesario iterar.

```python
article=soup.find_all('article')
for art in article:
    print(art.h1.text)
    break
    
```
