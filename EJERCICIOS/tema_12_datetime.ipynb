### 🧩 Ejercicio 2 – Día de la semana
Crea un `date` para tu cumpleaños y muestra qué día de la semana fue (usando `weekday()` y `isoweekday()`).  
> 🔍 *Reto:* Muestra el nombre del día en texto (“lunes”, “martes”, etc.).  
> 💡 *Habilidad:* interpretación de enteros devueltos por `weekday()`.

---

```python
def dia_semana(num_dia:int):
    dias:list=['Lunes','Martes','Miercoles','Jueves','Viernes','Sabado','Domingo']
    return dias[num_dia]
```

```python
fecha_cumple=datetime.date(2025,8,28)
print(dia_semana(fecha_cumple.weekday()))
```

### 🧩 Ejercicio 8 – Ajuste de hora con `timedelta`
A partir de un `datetime` de una reunión (por ejemplo, hoy a las 10:00 a.m.), calcula la hora de finalización si dura 2 horas y 45 minutos.  
> 💡 *Habilidad:* usar `timedelta(hours=..., minutes=...)`.

```python
reunion=datetime.datetime(2025,11,10,10,00,00)
tiempo_reunion=datetime.timedelta(hours=2,minutes=45)
finaliza_reunion=reunion+tiempo_reunion
print(finaliza_reunion.time())
```

### 🧩 Ejercicio 10 – Comparaciones
Crea dos `datetime` (por ejemplo, inicio y fin de un evento) y determina si el evento ya terminó y cuánto duró en horas.  
> 💡 *Habilidad:* comparación de `datetime` y resta.

```python
zona_horaria=pytz.UTC
registro=zona_horaria.localize(datetime.datetime(2025,11,10,19,00,00))
```

```python
date_inicio=zona_horaria.localize(datetime.datetime(2025,11,10,15,00,00))
date_final=zona_horaria.localize(datetime.datetime(2025,11,10,18,30,00))
duracion=(date_final-date_inicio)
print(duracion.total_seconds()/3600)
```

```python
if (date_inicio<registro<date_final):
    print('El evento esta en el aire')
else:
    print('El evento ha terminado')
```

### 🧩 Ejercicio 12 – Conversión de zona horaria
Convierte la hora UTC actual a:  
- Nueva York (`US/Eastern`)  
- Madrid (`Europe/Madrid`)  
- Tokio (`Asia/Tokyo`)  
> 💡 *Habilidad:* `.astimezone()` y comprensión del desplazamiento horario.

```python
hora_ny=hora_actual.astimezone(pytz.timezone('US/Eastern'))
print(hora_ny)
hora_ma=hora_actual.astimezone(pytz.timezone('Europe/Madrid'))
print(hora_ma)
hora_tk=hora_actual.astimezone(pytz.timezone('Asia/Tokyo'))
print(hora_tk)
```

```python
# Forma más recomendable ->Mas moderna
hora_utc = datetime.datetime.now(ZoneInfo("UTC"))
print("Hora UTC:", hora_utc)

# 2️⃣ Convertir a otras zonas horarias
hora_ny = hora_utc.astimezone(ZoneInfo("US/Eastern"))
hora_madrid = hora_utc.astimezone(ZoneInfo("Europe/Madrid"))
hora_tokio = hora_utc.astimezone(ZoneInfo("Asia/Tokyo"))

# 3️⃣ Mostrar resultados
print("Nueva York:", hora_ny)
print("Madrid:", hora_madrid)
print("Tokio:", hora_tokio)
```

### 🧩 Ejercicio 13 – Localización de hora
Simula que un usuario te da una hora “local” (sin zona horaria). Por ejemplo: “2025-11-10 14:30”.  
Localízala correctamente como hora de Buenos Aires (`America/Argentina/Buenos_Aires`).  
> 💡 *Habilidad:* `timezone.localize()` o `ZoneInfo` equivalente.

```python
hora_local=datetime.datetime(2025,11,10,14,30,00)
hora_argentina=hora_local.replace(tzinfo=ZoneInfo('America/Argentina/Buenos_Aires'))
print(hora_argentina)
```
