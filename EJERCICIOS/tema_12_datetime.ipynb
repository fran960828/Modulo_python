### 🧩 Ejercicio 18 – Registro de logs
Simula un sistema de logs que registra eventos con hora UTC (`datetime.now(tz=pytz.UTC)`).  
Cada vez que imprimas un evento, agrega una marca de tiempo ISO.  
> 🔍 *Reto:* Muestra además la hora local del usuario.  
> 💡 *Habilidad:* uso profesional de timestamps y UTC consistentes.

```python
import datetime
from zoneinfo import ZoneInfo

def registrar_horario_evento(evento: str, zona_local: str = 'Europe/Madrid', archivo_log: str = 'recursos/eventos.log'):
    # 1️⃣ Hora actual en UTC
    tiempo_UTC = datetime.datetime.now(ZoneInfo('UTC'))
    
    # 2️⃣ Conversión a hora local
    tiempo_local = tiempo_UTC.astimezone(ZoneInfo(zona_local))
    
    # 3️⃣ Formato ISO 8601 (profesional y legible)
    timestamp_UTC = tiempo_UTC.isoformat()
    timestamp_local = tiempo_local.isoformat()
    
    # 4️⃣ Crear línea de log
    linea_log = f"[UTC: {timestamp_UTC}] [Local: {timestamp_local}] Evento: {evento}\n"
    
    # 5️⃣ Mostrar en pantalla
    print(linea_log, end="")
    
    # 6️⃣ Guardar en archivo
    with open(archivo_log, "a", encoding="utf-8") as f:
        f.write(linea_log)
```

```python
registrar_horario_evento('iniciar Sesion')
```

### 🧩 Ejercicio 19 – Planificador internacional
Dado un evento programado en horario de Madrid (`Europe/Madrid`), muestra automáticamente la hora equivalente en Nueva York y Tokio.

### 🧩 Ejercicio 20 – Diferencia promedio de llegada
Tienes una lista de timestamps ISO de entregas realizadas por repartidores.  
Calcula la media del tiempo transcurrido entre cada entrega (en minutos).  
> 💡 *Habilidad:* parsing de múltiples fechas y cálculo estadístico con `timedelta`.

```python
from datetime import datetime
import numpy as np

# Lista original
lista_entregas = [
    "2025-11-10T10:05:00", "2025-11-10T11:25:00", "2025-11-10T11:10:00",
    "2025-11-10T10:50:00", "2025-11-10T11:45:00", "2025-11-10T11:31:00",
    "2025-11-10T10:18:00"
]

# 1️⃣ Convertir a datetime y ordenar
fechas = sorted(datetime.fromisoformat(f) for f in lista_entregas)

# 2️⃣ Calcular diferencias consecutivas
tiempos = [(fechas[i+1] - fechas[i]).total_seconds() for i in range(len(fechas)-1)]

# 3️⃣ Calcular media
media_segundos = np.mean(tiempos)
media_minutos = media_segundos / 60

# 4️⃣ Mostrar resultado
print(f"Tiempo promedio entre entregas: {media_minutos:.2f} minutos ({media_segundos:.1f} segundos)")
```

### 🧩 Ejercicio 21 – Validación de horarios laborales
Dada una hora local introducida por el usuario, determina si cae dentro del horario laboral (09:00–18:00) de su zona.  
> 🔍 *Reto:* Si no, muestra cuántas horas faltan para el próximo inicio de jornada.  
> 💡 *Habilidad:* manipulación condicional de `datetime.time`.

```python
from datetime import datetime, timedelta, date
from zoneinfo import ZoneInfo

hora_usuario = '2025-11-13 18:30'
hora_date = datetime.strptime(hora_usuario, '%Y-%m-%d %H:%M')
hora_local = hora_date.replace(tzinfo=ZoneInfo('Europe/Madrid'))

inicio_horario = datetime(2025, 11, 13, 9, 0, 0, tzinfo=ZoneInfo('Europe/Madrid'))
fin_horario = datetime(2025, 11, 13, 18, 0, 0, tzinfo=ZoneInfo('Europe/Madrid'))

if inicio_horario <= hora_local <= fin_horario:
    print("🟢 En horario laboral")
elif hora_local < inicio_horario:
    falta = inicio_horario - hora_local
    print(f"🕘 Aún no entras: faltan {falta.seconds // 3600} horas y {(falta.seconds % 3600) // 60} minutos")
else:
    # Después de fin de jornada
    # Calculamos el inicio del siguiente día laboral
    proximo_inicio = inicio_horario + timedelta(days=1)
    falta = proximo_inicio - hora_local
    print(f"🔴 Jornada terminada. Faltan {falta.seconds // 3600} horas y {(falta.seconds % 3600) // 60} minutos para el próximo inicio.")
```

### 🧩 Ejercicio 24 – Generador de reportes semanales
Genera una lista con las fechas de todos los **lunes** de un año determinado.  
> 🔍 *Reto:* Imprime también el número de semana ISO (`isocalendar()`).  
> 💡 *Habilidad:* iteración con `timedelta(days=7)` y uso de métodos ISO.

```python

paso = timedelta(days=1)
fecha_inicio = date(2025, 1, 1)
fecha_fin = date(2025, 12, 31)

lista_fechas = []

i = fecha_inicio
while i <= fecha_fin:
    if i.weekday() == 0:   # lunes
        lista_fechas.append(i)
    i += paso

lista_fechas
numero_semanas=date(2025,12,28).isocalendar().week
print(numero_semanas)
```

### 🧩 Ejercicio 25 – Dashboard de Fechas Inteligente
Crea un script que:
1. Muestre la fecha y hora actual en UTC y en tu zona local.  
2. Calcule automáticamente:
   - La fecha del próximo domingo.  
   - Cuántos días quedan para fin de año.  


```python
fecha_hoy = datetime.now()
paso = timedelta(days=1)
contador = 0

while fecha_hoy.weekday() != 6:  # 6 = domingo
    fecha_hoy += paso
    contador += 1

print(f'Faltan {contador} días para el domingo')

fecha_fin_ano=datetime(2025,12,31,0,0,0)
dias_fin_ano=fecha_fin_ano-fecha_hoy
print(f'Faltan {dias_fin_ano.days} dias hasta fin de año')
```
